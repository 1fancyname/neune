### Perceptron Components and Computations

In general the artificial neuron (perceptron) consists of the following components.



 ```{figure} images/Rosenblattperceptron.png
---

Perceptron Components  
Source: {cite}`wiki:rosenblatt_perceptron`

For a perceptron with $n$ inputs we have an input vector:
$$ 
x = \begin{pmatrix}
x_1 \\
x_2 \\
\vdots \\
x_n
\end{pmatrix}
$$

a weight vector:

$$ 
w = \begin{pmatrix}
w_1 \\
w_2 \\
\vdots \\
w_n
\end{pmatrix}
$$

an activation function $\varphi$ usually with a threshold $\theta$:

$$
\varphi =
f(z) =
\begin{cases}
1 & z \ge \theta \\
0 & z < \theta
\end{cases}
$$
